```python
# Data Preprocessing

import pandas as pd

# Load dataset
df = pd.read_csv("MajorFit (1).csv")

# Check missing values
print(df.isnull().sum())

# Handle missing values (if any)
df.fillna(df.mode().iloc[0], inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Dataset is already numerically encoded

# Check data types
print(df.dtypes)

# Final cleaned dataset preview
print(df.head())
```

```python
# Imports
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Split features and target using iloc
X = df.iloc[:, :-1].values
y = df.iloc[:, 10].values

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y
)

# Model 1: Logistic Regression
model1 = LogisticRegression()
model1.fit(X_train, y_train)

# Model 2: Decision Tree
model2 = DecisionTreeClassifier()
model2.fit(X_train, y_train)

# Model 3: Random Forest
model3 = RandomForestClassifier()
model3.fit(X_train, y_train)
```

### This section evaluates the three machine learning models using **Accuracy, Precision, Recall, F1-score**, and **Confusion Matrix**.

The evaluation is done on the test set only so the results show how well each model performs on unseen data.

```python
# Model Evaluation

from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Logistic Regression
print("Logistic Regression")

y_train_pred1 = model1.predict(X_train)
y_pred1 = model1.predict(X_test)

print('Training Accuracy : ', accuracy_score(y_train, y_train_pred1))
print('Test Accuracy : ', accuracy_score(y_test, y_pred1))

CR1 = classification_report(y_test, y_pred1)

print("Classification Report:")
print(CR1)

result1 = confusion_matrix(y_test, y_pred1)

sns.heatmap(result1, annot=True, fmt="g")
plt.xlabel('Predicted labels')
plt.ylabel('True labels')
plt.show()

# Decision Tree
print("Decision Tree")

y_train_pred2 = model2.predict(X_train)
y_pred2 = model2.predict(X_test)

print('Training Accuracy : ', accuracy_score(y_train, y_train_pred2))
print('Test Accuracy : ', accuracy_score(y_test, y_pred2))

CR2 = classification_report(y_test, y_pred2)

print("Classification Report:")
print(CR2)

result2 = confusion_matrix(y_test, y_pred2)

sns.heatmap(result2, annot=True, fmt="g")
plt.xlabel('Predicted labels')
plt.ylabel('True labels')
plt.show()

# Random Forest
print("Random Forest")

y_train_pred3 = model3.predict(X_train)
y_pred3 = model3.predict(X_test)

print('Training Accuracy : ', accuracy_score(y_train, y_train_pred3))
print('Test Accuracy : ', accuracy_score(y_test, y_pred3))

CR3 = classification_report(y_test, y_pred3)

print("Classification Report:")
print(CR3)

result3 = confusion_matrix(y_test, y_pred3)

sns.heatmap(result3, annot=True, fmt="g")
plt.xlabel('Predicted labels')
plt.ylabel('True labels')
plt.show()

evaluation_df = pd.DataFrame({
    'Model': ['Logistic Regression', 'Decision Tree', 'Random Forest'],

    'Accuracy': [
        accuracy_score(y_test, y_pred1),
        accuracy_score(y_test, y_pred2),
        accuracy_score(y_test, y_pred3)
    ],

    'Precision': [
        precision_score(y_test, y_pred1, average='weighted'),
        precision_score(y_test, y_pred2, average='weighted'),
        precision_score(y_test, y_pred3, average='weighted')
    ],

    'Recall': [
        recall_score(y_test, y_pred1, average='weighted'),
        recall_score(y_test, y_pred2, average='weighted'),
        recall_score(y_test, y_pred3, average='weighted')
    ],

    'F1-score': [
        f1_score(y_test, y_pred1, average='weighted'),
        f1_score(y_test, y_pred2, average='weighted'),
        f1_score(y_test, y_pred3, average='weighted')
    ]
})

evaluation_df = evaluation_df.sort_values(by='Accuracy', ascending=False)
```
